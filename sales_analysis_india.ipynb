# India-Localized Retail Sales Analysis

A guided exploratory analysis of a sample electronics sales dataset using Python and pandas. The visuals use an India-inspired palette and INR values.

> **Currency assumption:** The source notebook labels prices with `$`. This project uses a fixed illustrative conversion of ₹95 per US dollar. It is not a historical exchange rate or proof that these transactions took place in India. The dataset has no customer, city, or state columns, so this notebook does not infer customer or regional behavior.

## Questions

1. Is the supplied sales total consistent with quantity multiplied by unit price?
2. Which products and categories contribute the most revenue?
3. How does monthly revenue vary in this sample?
4. What are the average order value and units per order?

```python
from analysis import analyze_data, create_charts, format_crore, format_inr

df, monthly_revenue, product_revenue, category_revenue, quality, summary = analyze_data()
print(f"Loaded {summary['rows']:,} records covering {summary['date_start']} to {summary['date_end']}")
for metric, value in quality.items():
    print(f"{metric.replace('_', ' ').title()}: {value}")
```

## Data quality

The analysis removes exact duplicate rows, parses dates, checks required fields, confirms one row per order ID, requires positive quantities and prices, and verifies `TotalSales = Quantity × UnitPrice`. It stops with an error if a validation fails rather than silently presenting questionable totals.

```python
print('KEY METRICS — INR AT THE DOCUMENTED ILLUSTRATIVE RATE')
print(f"Total sales: {format_crore(summary['total_inr'])}")
print(f"Average order value: {format_inr(summary['average_order_value_inr'])}")
print(f"Average units per order: {summary['average_units_per_order']:.2f}")
print(f"Top product: {summary['top_product']} — {format_crore(summary['top_product_revenue_inr'])} ({summary['top_product_share']:.1%})")
print(f"Top category: {summary['top_category']} — {format_crore(summary['top_category_revenue_inr'])} ({summary['top_category_share']:.1%})")
print(f"Peak month: {summary['peak_month']} — {format_crore(summary['peak_month_revenue_inr'])}")
print(f"Lowest month: {summary['lowest_month']} — {format_crore(summary['lowest_month_revenue_inr'])}")
```

```python
product_table = product_revenue.rename('Revenue (INR)').to_frame()
product_table['Share of sales'] = product_table['Revenue (INR)'] / summary['total_inr']
print(product_table.to_string(formatters={
    'Revenue (INR)': lambda amount: format_inr(amount, 0),
    'Share of sales': lambda share: f'{share:.1%}',
}))
```

```python
monthly_table = monthly_revenue.rename('Revenue (INR)').to_frame()
print(monthly_table.to_string(formatters={'Revenue (INR)': lambda amount: format_inr(amount, 0)}))
```

```python
for chart_path in create_charts(monthly_revenue, product_revenue, category_revenue):
    print(f"Saved {chart_path.relative_to(chart_path.parents[1])}")
```
